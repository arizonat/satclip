```python
import numpy 
import pandas as pd
from pathlib import Path

root_dir = Path("/media/volume/satclip-datasets/Datasets/ts-satclip-s2-1M")

idx_df = pd.read_csv(root_dir / "full-index.csv")
# idx_df["ts"] =  pd.to_datetime(idx_df["ts"])

print(idx_df.head())
```

```python
# Create a balanced split
dts = pd.to_datetime(idx_df["ts"])
years = dts.dt.year.unique()

balanced = pd.DataFrame()

num_per_year = 200000

for year in years:
    idx_df_year = idx_df[dts.dt.year == year]
    samples = idx_df_year.sample(n=min(num_per_year, len(idx_df_year)))
    balanced = pd.concat([balanced, samples], ignore_index=True)

balanced.to_csv(root_dir / "index-balanced-1M.csv", index=False)

# Create mini splits
balanced.sample(n=10000).to_csv(root_dir / "index-10k.csv", index=False)
balanced.sample(n=100000).to_csv(root_dir / "index-100k.csv", index=False)
```
