```python
"""
NOTE: THIS SCRIPT MAKES A COPY OF THE DATASET, DO NOT USE IF RUNNING LOW ON STORAGE
"""
import pandas as pd
from pathlib import Path
import os

root_path = Path("/home/leca5365/Datasets/satclip-s2-temporal-55k/")
cutoff_date = "2025-01-01"
train_path = root_path.parent / "satclip-s2-temporal-55k-2021-2024"
test_path = root_path.parent / "satclip-s2-temporal-55k-2025-2026"

df = pd.read_csv(root_path / "index.csv")

# Cast ts column as datetime objects
df["dt"] = pd.to_datetime(df["ts"])
df_train = df[df["dt"] < cutoff_date]
df_test = df[df["dt"] >= cutoff_date]

# Make the corresponding train and test splits
os.makedirs(train_path, exist_ok=True)
os.makedirs(test_path, exist_ok=True)

df_train.to_csv(train_path / "index.csv", columns=["fn","id","lat","lon","ts"], index=False)
df_test.to_csv(test_path / "index.csv", columns=["fn","id","lat","lon","ts"], index=False)
```

```python
# Iterate through the images in each df and cp them into the new dataset folders
from shutil import copy2
from tqdm import tqdm

for dst_path, df_sub in zip([train_path, test_path], [df_train, df_test]):
    print("Copying image subset: ", dst_path)
    os.makedirs(dst_path / "images", exist_ok=True)

    for fn in tqdm(df_sub["fn"]):
        src = root_path / "images" / fn
        dst = dst_path / "images" / fn
        if src.exists():
            copy2(src, dst)
        else:
            print(f"Warning: {src} does not exist.")
```
